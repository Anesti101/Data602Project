```python
import boto3, os, json, pandas as pd
from dotenv import load_dotenv
from io import StringIO, BytesIO

load_dotenv()

s3 = boto3.client(
    's3',
    aws_access_key_id=os.getenv('AWS_ACCESS_KEY_ID'),
    aws_secret_access_key=os.getenv('AWS_SECRET_ACCESS_KEY'),
    region_name=os.getenv('AWS_REGION')
)

bucket = os.getenv('BUCKET_NAME')

# --- Sample one Academy CSV ---
csv_obj = s3.get_object(Bucket=bucket, Key='Academy/Business_20_2019-02-11.csv')
df_academy = pd.read_csv(StringIO(csv_obj['Body'].read().decode('utf-8')))
print("=== ACADEMY SAMPLE ===")
print(df_academy.shape)
print(df_academy.dtypes)
print(df_academy.head())

# --- Sample one Talent JSON ---
json_obj = s3.get_object(Bucket=bucket, Key='Talent/10383.json')
talent_record = json.loads(json_obj['Body'].read().decode('utf-8'))
print("\n=== TALENT SAMPLE ===")
print(json.dumps(talent_record, indent=2))
```

```python
# Debug: check credentials are loading from .env
print("Key loaded:", os.getenv("AWS_ACCESS_KEY_ID") is not None)
print("Secret loaded:", os.getenv("AWS_SECRET_ACCESS_KEY") is not None)
print("Region:", os.getenv("AWS_REGION"))
print("Bucket:", os.getenv("BUCKET_NAME"))

# Also verify you can reach S3 at all — list the bucket
response = s3.list_objects_v2(Bucket=bucket, MaxKeys=5)
for obj in response.get("Contents", []):
    print(obj["Key"])
```
